```python
#Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
#Load data
path="/content/drive/MyDrive/AXSOSACADEMY/03-IntermediateML/Week18/Data/credit_card.csv"
df=pd.read_csv(path)
df.info()
df.head()
```

```python
null = df.isnull().sum().sum()
null
```

```python
duplicates=df.duplicated().sum()
duplicates
```

```python
#Scale the data
scaler=StandardScaler()
scaled_df=scaler.fit_transform(df)
X_df=pd.DataFrame(scaled_df,columns=df.columns)
X_df.head()
```

```python
kmeans = KMeans(n_clusters=3,n_init='auto',random_state=42)
kmeans.fit(X_df)
```

```python
from scipy.spatial.distance import cdist
distance=cdist(X_df,kmeans.cluster_centers_,'euclidean')
distance.shape
```

```python
min_distances = np.min(distance,axis=1)
min_distances[:5]
```

```python
thresould = np.percentile(min_distances,99.6)
thresould
```

```python
filter_anomalies = min_distances > thresould
filter_anomalies.sum()
```

```python
idx_anomalies = X_df[filter_anomalies].index
idx_anomalies
```
